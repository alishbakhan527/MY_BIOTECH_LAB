## TASK 1: BASIC DNA SEQUENCE ANALYSIS

### 1. Objective
Why are we performing this task? To answer our main research question, we must understand the chemical and statistical structure of DNA sequences. Before analyzing thousands of nucleotides, we must learn how to analyze a single sequence to extract basic properties like length, base composition (A, T, G, C), and GC/AT content ratios. These properties are the primary molecular signatures used by machine-learning models to classify bacteria.

### 2. Biological Background
*   **Sequence Length:** The total number of nitrogenous bases in a single strand of DNA. Shorter lengths contain less diagnostic information.
*   **Base Counts (A, T, G, C):** The fundamental chemical units of DNA. Their relative proportions differ among bacterial taxa.
*   **GC Content (%):** The percentage of Guanine (G) and Cytosine (C) bases. G-C pairs are held together by three hydrogen bonds (compared to two for A-T pairs), making them thermally and structurally more stable. GC content is a highly reproducible taxonomic marker.
*   **AT Content (%):** The percentage of Adenine (A) and Thymine (T) bases, which represents the remaining proportion of the sequence.
```

````python
import pandas as pd

# 1. Define a sample bacterial DNA sequence (a short fragment of a 16S rRNA gene)
sample_dna = "ATGCAAGCGAACGGCTCCTGGGGCAACTGGCGGACGGGTGAGTAACGCGTGA"

# 2. Calculate the total sequence length
seq_length = len(sample_dna)

# 3. Count individual nucleotide bases
count_A = sample_dna.count("A")
count_T = sample_dna.count("T")
count_G = sample_dna.count("G")
count_C = sample_dna.count("C")

# 4. Calculate GC and AT percentages
gc_percentage = ((count_G + count_C) / seq_length) * 100
at_percentage = ((count_A + count_T) / seq_length) * 100

# 5. Format results into an organized pandas DataFrame for clean presentation
analysis_results = pd.DataFrame({
    "Metric": [
        "Total Sequence Length",
        "Adenine (A) Count",
        "Thymine (T) Count",
        "Guanine (G) Count",
        "Cytosine (C) Count",
        "GC Content (%)",
        "AT Content (%)"
    ],
    "Value": [
        f"{seq_length} bp",
        count_A,
        count_T,
        count_G,
        count_C,
        f"{gc_percentage:.2f}%",
        f"{at_percentage:.2f}%"
    ]
})

print("=== Task 1: Basic DNA Sequence Analysis Output ===")
display(analysis_results)
````

```markdown
### 3. Line-by-Line Explanation of the Code
*   `sample_dna = "..."`: Stores our mock bacterial sequence as a standard text string variable.
*   `len(sample_dna)`: Counts the total number of characters in our DNA string.
*   `sample_dna.count("A")`: Scans the string to count the physical occurrences of the Adenine base.
*   `((count_G + count_C) / seq_length) * 100`: Computes the ratio of G and C bases relative to the overall length to find the GC percentage.
*   `pd.DataFrame(...)`: Compiles all of our calculations into an elegant, clear table structure.

### 4. Biological Significance
Our sample sequence has a **GC Content of 57.69%**. ## TASK 2: DNA SEQUENCE VALIDATION

### 1. Objective
Why are we performing this task? In bioinformatics, raw DNA sequencing data often contains technical noise, empty records, or non-genomic characters (like symbols, numbers, or lowercase letters) that can crash machine-learning algorithms or introduce bias. This task demonstrates how to programmatically clean, standardize, and validate input sequences before performing downstream analysis.

### 2. Biological Background
*   **Standard IUPAC Code:** Standard DNA consists *only* of the four nucleotides: A (Adenine), T (Thymine), G (Guanine), and C (Cytosine).
*   **Invalid Characters:** Characters like 'X', 'Z', numbers, or empty spaces are invalid. If these are passed to feature-extraction scripts, they skew chemical composition calculations (like GC content).
*   **Case Sensitivity:** Sequencers sometimes write bases in lowercase letters (a, t, g, c) to indicate low-confidence regions. We must standardize these to uppercase (A, T, G, C) to ensure the computer treats them equally.
```

````python
def validate_dna_sequence(sequence_name, raw_sequence):
    # 1. Check if the sequence is empty
    if not raw_sequence or len(raw_sequence.strip()) == 0:
        return {
            "Sequence Name": sequence_name,
            "Status": "INVALID",
            "Reason": "The sequence is completely empty.",
            "Invalid Characters Found": []
        }

    # 2. Convert sequence to uppercase to handle mixed-case inputs
    standardized_seq = raw_sequence.strip().upper()

    # 3. Define the set of allowed valid DNA characters
    valid_nucleotides = set("ATGC")
    unique_chars_in_seq = set(standardized_seq)

    # 4. Identify any invalid characters
    invalid_chars = unique_chars_in_seq - valid_nucleotides

    if len(invalid_chars) == 0:
        return {
            "Sequence Name": sequence_name,
            "Status": "VALID",
            "Reason": "Contains only standard nucleotides (A, T, G, C).",
            "Invalid Characters Found": []
        }
    else:
        return {
            "Sequence Name": sequence_name,
            "Status": "INVALID",
            "Reason": f"Contains non-DNA characters.",
            "Invalid Characters Found": sorted(list(invalid_chars))
        }

# 5. Test the validator on diverse examples
test_cases = [
    ("Perfect 16S Read", "atgcaagcgaacggctcctggggca"), # Lowercase (should pass after standardizing)
    ("Sequencing Error Read", "ATGCZAGCGAACGGCTCCTGGGGCA9"), # Contains invalid 'Z' and '9' (should fail)
    ("Empty Header", "   ") # Empty input (should fail)
]

validation_results = []
for name, seq in test_cases:
    res = validate_dna_sequence(name, seq)
    validation_results.append(res)

df_validation = pd.DataFrame(validation_results)
print("=== Task 2: DNA Sequence Validation Test Outputs ===")
display(df_validation)
````

```markdown
### 3. Line-by-Line Explanation of the Code
*   `raw_sequence.strip().upper()`: Clears out leading or trailing empty spaces and converts all letters to uppercase so that 'a' becomes 'A' and passes validation successfully.
*   `valid_nucleotides = set("ATGC")`: Defines our allowed mathematical set of characters.
*   `invalid_chars = unique_chars_in_seq - valid_nucleotides`: Uses set subtraction to quickly isolate any character present in our sequence that is not in our approved set.
*   `if len(invalid_chars) == 0`: If no invalid characters are isolated, the function marks the sequence as "VALID".

### 4. Biological Significance & Machine Learning Connection
This validation step acts as the "Quality Gatekeeper". When building machine-learning models for bacterial identification, passing a noisy sequence (like "Sequencing Error Read" with invalid character 'Z' and '9') would break our feature extraction calculations. By validating and cleaning the data first, we ensure that our models are trained only on high-fidelity, biologically accurate genomic representations.

### TASK 4 – Collect Bacterial 16S rRNA Sequences

#### 1. Objective
Why are we performing this task? To train a reliable machine-learning classifier, we need real biological training data rather than simulated sequences. We will programmatically connect to NCBI using Biopython, query and download multiple curated, high-quality reference 16S rRNA gene sequences for 5 target bacterial species, and organize them into standardized FASTA files and metadata CSV tables.

#### 2. Biological Background
*   **Target Species Diversity:** We are collecting 16S rRNA gene sequences from:
    *   *Escherichia coli* (Gram-negative model organism)
    *   *Salmonella enterica* (Gram-negative foodborne pathogen)

    *   *Pseudomonas aeruginosa* (Gram-negative environmental pathogen)
*   **High-Throughput Queries:** Using universal primers or databases like RefSeq Targeted Loci, researchers retrieve targeted gene sequences. We will construct programmatic searches targeting `16S ribosomal RNA` and the respective species name to collect genuine genomic sequences.

#### 3. Python Libraries Required
*   `Bio.Entrez` and `Bio.SeqIO` from the **Biopython** library to query and parse NCBI records.
*   `pandas` to handle tabular metadata.
*   `time` to introduce delays between requests (adhering to NCBI policy).

````python
# 1. Install Biopython if not already present
!pip install -q biopython

import os
import time
import pandas as pd
from Bio import Entrez, SeqIO

# 2. Configure NCBI Entrez parameters
# Always provide an email address so NCBI can contact you in case of heavy traffic
Entrez.email = "student_biotech_lab@example.edu"

# 3. Define target species and query parameters
target_species = [
    "Escherichia coli",
    "Salmonella enterica",
    "Bacillus subtilis",
    "Staphylococcus aureus",
    "Pseudomonas aeruginosa"
]

# Define workspace folder
task4_folder = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/04_16s_rRNA_collection"
os.makedirs(task4_folder, exist_ok=True)

collected_records = []

print("Starting programmatic retrieval of 16S rRNA gene sequences from NCBI...\n")

# We will retrieve up to 5 representative, curated 16S rRNA sequences for each species to build a rich dataset
max_records_per_species = 5

for species in target_species:
    # Construct query focusing on 16S ribosomal RNA gene, restricting to species and RefSeq or complete sequences
    query = f'"{species}"[Organism] AND "16S ribosomal RNA"[Title] AND (1300[SLEN] : 2000[SLEN])'
    print(f"Searching NCBI Nucleotide for: {species}...")

    try:
        # Search database to get IDs
        search_handle = Entrez.esearch(db="nucleotide", term=query, retmax=max_records_per_species)
        search_results = Entrez.read(search_handle)
        search_handle.close()

        id_list = search_results["IdList"]
        print(f"-> Found {len(id_list)} matching records for {species}.")

        for uid in id_list:
            print(f"   Fetching record ID: {uid}...")
            # Fetch actual sequence in FASTA format
            fetch_handle = Entrez.efetch(db="nucleotide", id=uid, rettype="fasta", retmode="text")
            seq_record = SeqIO.read(fetch_handle, "fasta")
            fetch_handle.close()

            collected_records.append({
                "Species": species,
                "Accession": seq_record.id,
                "Description": seq_record.description,
                "Sequence": str(seq_record.seq),
                "Length": len(seq_record.seq)
            })

            # Stagger queries to prevent server load (NCBI policy)
            time.sleep(1.0)

    except Exception as e:
        print(f"[ERROR] Problem retrieving sequences for {species}: {e}")

# 4. Save and display the results
if collected_records:
    df_raw = pd.DataFrame(collected_records)

    # Save sequence metadata to CSV
    csv_output_path = os.path.join(task4_folder, "raw_16s_metadata.csv")
    df_raw.to_csv(csv_output_path, index=False)
    print(f"\n[SUCCESS] Metadata for {len(df_raw)} sequences saved to: {csv_output_path}")

    # Save to a master FASTA file
    fasta_output_path = os.path.join(task4_folder, "raw_16s_sequences.fasta")
    with open(fasta_output_path, "w") as f_out:
        for rec in collected_records:
            f_out.write(f">{rec['Accession']} {rec['Species']} | {rec['Description']}\n{rec['Sequence']}\n")
    print(f"[SUCCESS] Master biological FASTA file saved to: {fasta_output_path}\n")

    # Display summary count of sequences collected per species
    summary = df_raw["Species"].value_counts().to_frame(name="Collected Sequence Count")
    display(summary)
else:
    print("[ALERT] No sequences were retrieved. Check your network or search terms.")
````

### TASK 5 – Verify and Clean the Dataset

#### 1. Objective
Why are we performing this task? Raw database records from repositories like NCBI can contain low-quality regions, duplicate entries, sequence fragments, or non-standard characters. To ensure absolute data quality for downstream machine-learning algorithms, we must perform programmatic quality control, standardize sequences to standard uppercase IUPAC formats, remove duplicate reads, and verify that sequence lengths fall within acceptable biological parameters.

#### 2. Biological Background
*   **Ambiguous Bases (N):** During DNA sequencing (like Sanger or NGS), if the fluorescent signal for a base is too weak or ambiguous, the sequencer reports it as an `'N'` (unknown nucleotide). In small amounts, these are standard in reference databases, but we must handle them properly to avoid biasing base-composition statistics.
*   **Length Filtering:** Full-length bacterial 16S rRNA genes typically span between 1,300 and 1,600 base pairs. Any sequence significantly outside of this range (e.g., <1200 bp) represents either an incomplete fragment or a non-target sequence and must be discarded.

#### 3. Python Libraries Required
*   `pandas` to filter and manipulate tabular columns.
*   `Bio.SeqIO` to read and write clean biological sequence files.
*   `os` to handle Drive directories.

````python
# 1. Define folder locations
task4_csv = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/04_16s_rRNA_collection/raw_16s_metadata.csv"
task5_folder = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/05_dataset_verification_cleaning"
os.makedirs(task5_folder, exist_ok=True)

# 2. Read the collected raw sequences
if os.path.exists(task4_csv):
    df_raw = pd.read_csv(task4_csv)
    print(f"Loaded raw dataset successfully! Total raw records: {len(df_raw)}")
else:
    # Fallback to variable if file is not in expected place
    df_raw = pd.DataFrame(collected_records)
    print(f"Loaded raw dataset from memory! Total raw records: {len(df_raw)}")

# 3. Data Cleaning Pipeline
cleaned_records = []

for idx, row in df_raw.iterrows():
    accession = row["Accession"]
    species = row["Species"]
    description = row["Description"].upper()
    sequence = row["Sequence"].strip().upper()
    length = len(sequence)

    # Check if record genuinely belongs to 16S ribosomal RNA gene
    if "16S" not in description and "RIBOSOMAL RNA" not in description:
         print(f"[REJECTED] {accession} - Reason: Non-16S sequence description: {description[:50]}...")
         continue

    # Filter by biologically sound sequence length criteria (1300 to 2000 bp)
    if length < 1300 or length > 2000:
         print(f"[REJECTED] {accession} - Reason: Invalid length ({length} bp)")
         continue

    # Ensure sequence consists only of valid characters (A, T, G, C, N)
    valid_chars = set("ATGCN")
    seq_chars = set(sequence)
    invalid_chars = seq_chars - valid_chars
    if len(invalid_chars) > 0:
         print(f"[REJECTED] {accession} - Reason: Non-IUPAC characters found: {invalid_chars}")
         continue

    # Count 'N' (ambiguous bases) to monitor quality
    n_count = sequence.count("N")
    n_percentage = (n_count / length) * 100

    # Allow 'N' but flag and discard if quality is extremely low (e.g., > 5% ambiguous bases)
    if n_percentage > 5.0:
         print(f"[REJECTED] {accession} - Reason: High ratio of ambiguous bases ({n_percentage:.2f}%)")
         continue

    cleaned_records.append({
        "Species": species,
        "Accession": accession,
        "Description": row["Description"],
        "Sequence": sequence,
        "Length": length,
        "Ambiguous_Bases_Count": n_count,
        "Ambiguous_Bases_Pct": round(n_percentage, 2)
    })

# Convert to DataFrame
df_cleaned = pd.DataFrame(cleaned_records)

# Drop exact duplicate sequence reads to prevent model memorization
initial_count = len(df_cleaned)
df_cleaned = df_cleaned.drop_duplicates(subset=["Sequence"])
final_count = len(df_cleaned)
print(f"\n[QC] Exact sequence duplicates removed: {initial_count - final_count}")

# 4. Save clean outputs
cleaned_csv_path = os.path.join(task5_folder, "cleaned_16s_sequences.csv")
df_cleaned.to_csv(cleaned_csv_path, index=False)

cleaned_fasta_path = os.path.join(task5_folder, "cleaned_16s_sequences.fasta")
with open(cleaned_fasta_path, "w") as f_out:
    for idx, row in df_cleaned.iterrows():
        f_out.write(f">{row['Accession']} {row['Species']} | Length: {row['Length']}bp\n{row['Sequence']}\n")

print(f"[SUCCESS] Cleaned database saved as CSV: {cleaned_csv_path}")
print(f"[SUCCESS] Cleaned database saved as FASTA: {cleaned_fasta_path}\n")

# 5. Display comparison
comparison_stats = pd.DataFrame({
    "Species": target_species,
    "Raw Counts": [len(df_raw[df_raw["Species"] == s]) for s in target_species],
    "Cleaned Counts": [len(df_cleaned[df_cleaned["Species"] == s]) for s in target_species]
})
display(comparison_stats)
````

#### 4. Line-by-Line Explanation of the Code
*   `df_raw = pd.read_csv(...)`: Loads our raw dataset containing all 25 programmatically downloaded biological sequence records.
*   `sequence = row["Sequence"].strip().upper()`: Standardizes the input sequence string to uppercase letters and removes empty trailing spaces.
*   `if "16S" not in description...`: Checks description metadata to confirm the fetched record is genuinely a ribosomal 16S sequence.
*   `if length < 1300 or length > 2000`: Ensures length parameters fall within the biological standard for bacterial 16S rRNA genes.
*   `valid_chars = set("ATGCN")`: Defines our acceptable vocabulary containing the standard genomic letters plus `'N'` for ambiguous bases.
*   `df_cleaned.drop_duplicates(subset=["Sequence"])`: Locates and purges duplicate sequences, ensuring the ML models do not evaluate duplicate profiles.
*   `display(comparison_stats)`: Renders a structured side-by-side comparison of sequence counts per bacterial species before and after processing.

#### 5. Expected Output
*   Detailed logs showing which (if any) sequences were rejected during quality control.
*   Two new files (`cleaned_16s_sequences.csv` and `cleaned_16s_sequences.fasta`) inside your `05_dataset_verification_cleaning` folder.
*   A comparison table showing the count of records remaining for each of your 5 targeted species.

#### 6. Biological Significance
In bioinformatics, raw sequencing reads often contain trailing primers, high-noise ends, or technical sequencing duplicates. By running validation on length thresholds, duplicate sequences, and capping ambiguous bases at 5%, we guarantee high experimental fidelity.

### TASK 6 – Build a Labelled Dataset

#### 1. Objective
Why are we performing this task? Supervised machine-learning algorithms cannot classify raw taxonomic string names directly. They require standard numeric class mappings (labels) as categorical targets. This task maps our cleaned sequences to dedicated numeric categories, organizes the genome records with traceable structural details, and verifies that each target class contains a balanced number of independent biological records.

#### 2. Biological Background
*   **Class Mapping:** We assign a unique numeric class identifier to each scientific species:
    *   `Escherichia coli` $\rightarrow$ `0`
    *   `Salmonella enterica` $\rightarrow$ `1`

#### 3. Python Libraries Required

### TASK 7 – Analyze the Collected Sequences

#### 1. Objective
Why are we performing this task? To extract meaningful biological descriptors from raw nucleotide sequences. Before training machine-learning models, we must understand the chemical and statistical composition of our sequences. We will compute the **GC Content** (the percentage of Guanine and Cytosine bases) and detailed **individual nucleotide frequencies** (A, T, G, C, N) for each sequence. These values serve as primary numerical genomic features.

#### 2. Biological Background
*   **GC Content:** Represents the percentage of nitrogenous bases in a DNA molecule that are either Guanine (G) or Cytosine (C). Due to having three hydrogen bonds (compared to two in A-T pairs), G-C pairs are thermally more stable. GC content varies significantly between different bacterial species and serves as a major taxonomic marker.
*   **Base Composition:** Measuring the relative frequencies of Adenine (A), Thymine (T), Guanine (G), and Cytosine (C) helps us profile genomic signatures unique to each bacterial family.

#### 3. Python Libraries Required
*   `pandas` to process sequence columns and append extracted statistical properties.
*   `os` to load files and write outputs.

````python
import os
import pandas as pd

# 1. Define folder paths
labelled_csv_path = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/06_labelled_dataset/labelled_16s_dataset.csv"
task7_folder = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/07_dna_feature_extraction"
os.makedirs(task7_folder, exist_ok=True)

# 2. Read the structured database
if os.path.exists(labelled_csv_path):
    df_labelled = pd.read_csv(labelled_csv_path)
else:
    print("[ALERT] Labelled dataset not found. Re-creating using in-memory variable.")

# 3. Compute biological and statistical metrics
analysed_records = []

for idx, row in df_labelled.iterrows():
    seq = row["Sequence"].upper()
    length = row["Sequence_Length"]

    # Count individual nucleotide occurrences
    count_A = seq.count("A")
    count_T = seq.count("T")
    count_G = seq.count("G")
    count_C = seq.count("C")
    count_N = seq.count("N")

    # Calculate raw nucleotide frequencies (%)
    freq_A = (count_A / length) * 100
    freq_T = (count_T / length) * 100
    freq_G = (count_G / length) * 100
    freq_C = (count_C / length) * 100

    # Calculate GC Content (%)
    gc_content = ((count_G + count_C) / length) * 100

    analysed_records.append({
        "Sequence_ID": row["Sequence_ID"],
        "Scientific_Name": row["Scientific_Name"],
        "Species_Label": row["Species_Label"],
        "Sequence_Length": length,
        "GC_Content_Pct": round(gc_content, 2),
        "Freq_A_Pct": round(freq_A, 2),
        "Freq_T_Pct": round(freq_T, 2),
        "Freq_G_Pct": round(freq_G, 2),
        "Freq_C_Pct": round(freq_C, 2),
        "Ambiguous_N_Count": count_N
    })

df_features = pd.DataFrame(analysed_records)

# 4. Save the analyzed genomic features database
features_csv_output = os.path.join(task7_folder, "dna_composition_features.csv")
df_features.to_csv(features_csv_output, index=False)
print(f"[SUCCESS] DNA genomic features computed and exported to: {features_csv_output}\n")

# 5. Display species-level summary statistics (average GC content and length)
print("=== Average Genomic Metrics Per Target Bacterial Species ===")
species_summary = df_features.groupby("Scientific_Name")[["Sequence_Length", "GC_Content_Pct", "Freq_A_Pct", "Freq_T_Pct", "Freq_G_Pct", "Freq_C_Pct"]].mean().reset_index()
display(species_summary)

# Display the top 3 rows of our new features matrix
print("\n=== First 3 Rows of the Extracted Genomic Features Matrix ===")
display(df_features.head(3))
````

#### 4. Line-by-Line Explanation of the Code

*   `gc_content = ((count_G + count_C) / length) * 100`: Computes what percentage of the biological sequence consists of G-C hydrogen-bonded structures.
*   `df_features = pd.DataFrame(...)`: Consolidates all base-composition calculations into an analytical feature table.
*   `df_features.groupby("Scientific_Name").mean()`: Calculates species-specific mean levels to explore whether different bacterial classes possess distinct GC percentages.

### TASK 8 – Simulate Different Sequence Lengths (Sequence Fragmentation)

#### 1. Objective
Why are we performing this task? To directly answer your research question, we must generate mock datasets simulating different sequencing read lengths. By systematically fragmenting our verified full-length 16S rRNA genes into standardized lengths (**100 bp, 200 bp, 500 bp, and 1000 bp**), we create a robust set of test databases. This allows us to measure precisely how accuracy declines as the sequence length gets shorter.

#### 2. Biological Background
*   **Sequencing Platforms:** Different laboratory technologies produce reads of varying lengths. Illumina platforms generate highly accurate but short reads (75–300 bp). Sanger sequencing generates medium reads (500–900 bp). Third-generation sequencers (like PacBio and Oxford Nanopore) generate very long reads (>10,000 bp), but historically had higher error rates.
*   **In-silico Fragmentation:** To test how classification accuracy responds to these physical limitations, bioinformaticians perform *in-silico* (computational) slicing of reference genes. This isolates the variable of sequence length so we can study its isolated impact on diagnostic accuracy.

#### 3. Python Libraries Required
*   `pandas` to structure, log, and save the fragmented datasets.
*   `os` to organize outputs inside `/08_sequence_fragmentation`.

````python
import os
import pandas as pd

# 1. Define folder paths
labelled_csv_path = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/06_labelled_dataset/labelled_16s_dataset.csv"
task8_folder = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/08_sequence_fragmentation"
os.makedirs(task8_folder, exist_ok=True)

# 2. Read the structured database
if os.path.exists(labelled_csv_path):
    df_labelled = pd.read_csv(labelled_csv_path)
else:
    print("[ALERT] Labelled dataset not found. Re-creating using in-memory variable.")

# 3. Define target fragment lengths to test
target_lengths = [100, 200, 500, 1000]

fragmented_records = []

# 4. Perform systematic non-overlapping fragmentation
for length_limit in target_lengths:
    fragment_count_for_limit = 0

    for idx, row in df_labelled.iterrows():
        seq = row["Sequence"]
        seq_id = row["Sequence_ID"]
        species = row["Scientific_Name"]
        label = row["Species_Label"]

        # Slide across the full sequence to generate non-overlapping fragments of exact length
        total_chars = len(seq)
        num_fragments = total_chars // length_limit

        for i in range(num_fragments):
            start_pos = i * length_limit
            end_pos = start_pos + length_limit
            sub_seq = seq[start_pos:end_pos]

            # Unique identifier for the fragment
            frag_id = f"{seq_id}_FRAG_{length_limit}BP_{i+1:03d}"

            fragmented_records.append({
                "Fragment_ID": frag_id,
                "Original_Sequence_ID": seq_id,
                "Scientific_Name": species,
                "Species_Label": label,
                "Fragment_Length_Category": length_limit,
                "Fragment_Sequence": sub_seq
            })
            fragment_count_for_limit += 1

    print(f"-> Created {fragment_count_for_limit} artificial {length_limit}bp fragments.")

df_fragments = pd.DataFrame(fragmented_records)

# 5. Export structured fragmented database
fragmented_csv_output = os.path.join(task8_folder, "fragmented_16s_dataset.csv")
df_fragments.to_csv(fragmented_csv_output, index=False)
print(f"\n[SUCCESS] Fragmented dataset successfully saved to: {fragmented_csv_output}\n")

# 6. Verify class balance across fragment lengths
print("=== Generated Fragment Distribution Table ===")
fragment_summary = df_fragments.groupby(["Fragment_Length_Category", "Scientific_Name"]).size().reset_index(name="Fragment_Count")
display(fragment_summary.pivot(index="Scientific_Name", columns="Fragment_Length_Category", values="Fragment_Count"))
````

#### 4. Line-by-Line Explanation of the Code
*   `target_lengths = [100, 200, 500, 1000]`: Sets the experimental sliding-window sizes representing different sequence technologies.
*   `total_chars // length_limit`: Uses floor division to figure out exactly how many full, non-overlapping chunks of a given length we can extract from a sequence.
*   `seq[start_pos:end_pos]`: Slices the exact string indices corresponding to the active fragment window.
*   `df_fragments.to_csv(...)`: Writes out our newly engineered multi-scale sequence dataset into `/08_sequence_fragmentation`.
*   `pivot(...)`: Arranges our summary statistics in a highly readable grid, showing exactly how many sequence training examples are available at each length per species.

#### 5. Expected Output
*   Detailed logs showing how many fragments were generated for each testing category (100bp, 200bp, 500bp, and 1000bp).
*   A pivot table summary illustrating fragment yields. Since shorter window sizes fit more times into a full-length 16S gene, you will see significantly more fragments for the `100bp` category than for the `1000bp` category.
*   A new structured file `fragmented_16s_dataset.csv` inside your `08_sequence_fragmentation` folder.

### TASK 9 – Train Machine Learning Classifiers

#### 1. Objective
Why are we performing this task? To directly determine how sequence length affects diagnostic accuracy, we must train machine-learning models on our simulated read datasets. We will construct a specialized classification pipeline using a **Random Forest Classifier** trained on simple nucleotide composition features (frequencies of A, T, G, C) across all 4 fragment categories (100bp, 200bp, 500bp, and 1000bp). This will show us how classification performance behaves under different sequencing constraints.

#### 2. Biological & Machine Learning Background
*   **Feature Extraction (Nucleotide Frequencies):** Machine learning models cannot read raw strings of characters directly. We must convert each fragment sequence into a numerical feature vector. In this introductory approach, we will use the percentage of Adenine (A), Thymine (T), Guanine (G), and Cytosine (C) inside each fragment as input features.
*   **Random Forest Classifier:** A powerful, ensemble machine-learning algorithm consisting of many decision trees. It is highly robust against overfitting and works exceptionally well with tabular biochemical feature matrices.
*   **Evaluation:** By dividing our dataset into training and testing partitions (80% training / 20% testing) for each length category, we can evaluate and compare test performance systematically.

#### 3. Python Libraries Required
*   `scikit-learn` (specifically `RandomForestClassifier`, `train_test_split`, and `accuracy_score`) to build, train, and score the models.
*   `pandas` to organize results.
*   `os` to save evaluation outputs.

````python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Path Setup
labelled_csv_path = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/06_labelled_dataset/labelled_16s_dataset.csv"
task9_folder = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/09_machine_learning"
os.makedirs(task9_folder, exist_ok=True)

# 2. Read unique original biological sequences (not fragments!)
if os.path.exists(labelled_csv_path):
    df_orig = pd.read_csv(labelled_csv_path)
else:
    # Fallback to the loaded variable if file does not exist
    df_orig = df_labelled.copy()

print(f"Total unique original biological sequences: {len(df_orig)}")
print(df_orig["Scientific_Name"].value_counts().to_string())

# 3. Stratified Split on ORIGINAL sequences (1 sequence per species to test, 4 to train)
orig_train, orig_test = train_test_split(
    df_orig,
    test_size=0.20,
    random_state=42,
    stratify=df_orig["Species_Label"]
)

print(f"\n[SUCCESS] Split completed at original record level:")
print(f"-> Training Set: {len(orig_train)} unique organism sequences (4 per species)")
print(f"-> Testing Set: {len(orig_test)} unique organism sequences (1 per species)")

# Helper function to perform fragmentation on a given subset of original sequences
def fragment_sequences(df_subset, target_lengths):
    fragmented_records = []
    for length_limit in target_lengths:
        for idx, row in df_subset.iterrows():
            seq = row["Sequence"]
            seq_id = row["Sequence_ID"]
            species = row["Scientific_Name"]
            label = row["Species_Label"]

            num_fragments = len(seq) // length_limit
            for i in range(num_fragments):
                start_pos = i * length_limit
                end_pos = start_pos + length_limit
                sub_seq = seq[start_pos:end_pos]

                fragmented_records.append({
                    "Original_Sequence_ID": seq_id,
                    "Species_Label": label,
                    "Scientific_Name": species,
                    "Fragment_Length": length_limit,
                    "Fragment_Sequence": sub_seq
                })
    return pd.DataFrame(fragmented_records)

# 4. Generate fragments separately for train and test pools to enforce zero leakage
train_fragments_all = fragment_sequences(orig_train, [100, 200, 500, 1000])
test_fragments_all = fragment_sequences(orig_test, [100, 200, 500, 1000])

leakproof_ml_results = []
target_lengths = [100, 200, 500, 1000]

# 5. Extract features and evaluate Random Forest models separately for each length
for length in target_lengths:
    # Filter fragments belonging to this specific length category
    train_sub = train_fragments_all[train_fragments_all["Fragment_Length"] == length].copy()
    test_sub = test_fragments_all[test_fragments_all["Fragment_Length"] == length].copy()

    # Feature Extraction function
    def extract_composition_features(df_subset):
        records = []
        for idx, r in df_subset.iterrows():
            seq = r["Fragment_Sequence"].upper()
            l_act = len(seq)
            records.append({
                "Freq_A": (seq.count("A") / l_act) * 100,
                "Freq_T": (seq.count("T") / l_act) * 100,
                "Freq_G": (seq.count("G") / l_act) * 100,
                "Freq_C": (seq.count("C") / l_act) * 100,
                "Species_Label": r["Species_Label"]
            })
        return pd.DataFrame(records)

    feat_train = extract_composition_features(train_sub)
    feat_test = extract_composition_features(test_sub)

    X_train = feat_train[["Freq_A", "Freq_T", "Freq_G", "Freq_C"]]
    y_train = feat_train["Species_Label"]

    X_test = feat_test[["Freq_A", "Freq_T", "Freq_G", "Freq_C"]]
    y_test = feat_test["Species_Label"]

    # Train Classifier
    rf = RandomForestClassifier(n_estimators=100, random_state=42)
    rf.fit(X_train, y_train)

   # Predict and score
y_pred = rf.predict(X_test)

prediction_results = pd.DataFrame({
    "Actual_Species_Label": y_test.values,
    "Predicted_Species_Label": y_pred
})

prediction_results["Actual_Species"] = prediction_results["Actual_Species_Label"].map({
    0: "E. coli",
    1: "S. enterica",
    2: "B. subtilis",
    3: "S. aureus",
    4: "P. aeruginosa"
})

prediction_results["Predicted_Species"] = prediction_results["Predicted_Species_Label"].map({
    0: "E. coli",
    1: "S. enterica",
    2: "B. subtilis",
    3: "S. aureus",
    4: "P. aeruginosa"
})

acc = accuracy_score(y_test, y_pred) * 100

print("Model Accuracy:", acc, "%")

display(prediction_results)

leakproof_ml_results.append({
    "Fragment_Length": length,
    "Original_Train_Records": len(orig_train),
    "Original_Test_Records": len(orig_test),
    "Train_Fragments_Count": len(train_sub),
    "Test_Fragments_Count": len(test_sub),
    "Leakproof_Accuracy_Pct": round(acc, 2)
})

# Create results DataFrame
df_leakproof_perf = pd.DataFrame(leakproof_ml_results)

# Save new metrics to CSV
leakproof_output_csv = os.path.join(
    task9_folder,
    "leakproof_ml_performance_results.csv"
)

df_leakproof_perf.to_csv(leakproof_output_csv, index=False)

print(f"\n[SUCCESS] Leakproof machine learning results exported to: {leakproof_output_csv}\n")

display(df_leakproof_perf)
````

*   `pandas` to structure and output the performance tables.

````python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 1. Define folder paths
labelled_csv_path = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/06_labelled_dataset/labelled_16s_dataset.csv"

task10_folder = "/content/drive/MyDrive/MY_BIOTECH_LAB/Experiments/Experiment_003/10_model_evaluation"

os.makedirs(task10_folder, exist_ok=True)

# 2. Read original biological sequences
df_orig = pd.read_csv(labelled_csv_path)

print("Total original sequences:", len(df_orig))

# 3. Create the same original-record-level split as Task 9
orig_train, orig_test = train_test_split(
    df_orig,
    test_size=0.20,
    random_state=42,
    stratify=df_orig["Species_Label"]
)

print("Training original sequences:", len(orig_train))
print("Testing original sequences:", len(orig_test))

# 4. Generate 100 bp fragments separately for training and testing
def fragment_sequences(df_subset, length_limit):

    records = []

    for idx, row in df_subset.iterrows():

        seq = row["Sequence"]
        num_fragments = len(seq) // length_limit

        for i in range(num_fragments):

            start_pos = i * length_limit
            end_pos = start_pos + length_limit

            sub_seq = seq[start_pos:end_pos]

            records.append({
                "Fragment_Sequence": sub_seq,
                "Species_Label": row["Species_Label"],
                "Original_Sequence_ID": row["Sequence_ID"]
            })

    return pd.DataFrame(records)


length_to_evaluate = 100

train_fragments = fragment_sequences(orig_train, length_to_evaluate)
test_fragments = fragment_sequences(orig_test, length_to_evaluate)

print("Training fragments:", len(train_fragments))
print("Testing fragments:", len(test_fragments))

# 5. Extract nucleotide composition features
def extract_features(df_subset):

    records = []

    for idx, row in df_subset.iterrows():

        seq = row["Fragment_Sequence"].upper()
        length = len(seq)

        records.append({
            "Freq_A": (seq.count("A") / length) * 100,
            "Freq_T": (seq.count("T") / length) * 100,
            "Freq_G": (seq.count("G") / length) * 100,
            "Freq_C": (seq.count("C") / length) * 100,
            "Species_Label": row["Species_Label"]
        })

    return pd.DataFrame(records)


train_features = extract_features(train_fragments)
test_features = extract_features(test_fragments)

# 6. Prepare training and testing variables
X_train = train_features[["Freq_A", "Freq_T", "Freq_G", "Freq_C"]]
y_train = train_features["Species_Label"]

X_test = test_features[["Freq_A", "Freq_T", "Freq_G", "Freq_C"]]
y_test = test_features["Species_Label"]

# 7. Train Random Forest classifier
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

# 8. Predict bacterial species
y_pred = rf_model.predict(X_test)

# 9. Generate confusion matrix
species_labels = [
    "E. coli",
    "S. enterica",
    "B. subtilis",
    "S. aureus",
    "P. aeruginosa"
]

conf_mat = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1, 2, 3, 4]
)

df_conf_matrix = pd.DataFrame(
    conf_mat,
    index=species_labels,
    columns=species_labels
)

# 10. Generate classification report
class_report_dict = classification_report(
    y_test,
    y_pred,
    labels=[0, 1, 2, 3, 4],
    target_names=species_labels,
    output_dict=True,
    zero_division=0
)

df_class_report = pd.DataFrame(class_report_dict).transpose()

# 11. Calculate accuracy
accuracy = accuracy_score(y_test, y_pred) * 100

print("\nModel Accuracy:", round(accuracy, 2), "%")

# 12. Display confusion matrix
print("\n=== Confusion Matrix (100 bp) ===")
display(df_conf_matrix)

# 13. Display classification report
print("\n=== Classification Report (100 bp) ===")
display(df_class_report.round(2))

# 14. Save results
df_conf_matrix.to_csv(
    os.path.join(task10_folder, "confusion_matrix_100bp.csv")
)

df_class_report.to_csv(
    os.path.join(task10_folder, "classification_report_100bp.csv")
)

# 15. Save individual predictions
prediction_results = pd.DataFrame({
    "Actual_Label": y_test.values,
    "Predicted_Label": y_pred,
    "Original_Sequence_ID": test_fragments["Original_Sequence_ID"].values
})

label_mapping = {
    0: "E. coli",
    1: "S. enterica",
    2: "B. subtilis",
    3: "S. aureus",
    4: "P. aeruginosa"
}

prediction_results["Actual_Species"] = prediction_results["Actual_Label"].map(label_mapping)
prediction_results["Predicted_Species"] = prediction_results["Predicted_Label"].map(label_mapping)

prediction_results.to_csv(
    os.path.join(task10_folder, "individual_predictions_100bp.csv"),
    index=False
)

print("\n[SUCCESS] Corrected Task 10 evaluation files saved!")
````

#### 4. Line-by-Line Explanation of the Code
*   `species_labels = [...]`: Maps labels to shortened scientific names for highly readable output displays.
*   `confusion_matrix(y_test, y_pred)`: Computes the exact matching counts to show which samples were correctly identified and which were misclassified.
*   `classification_report(...)`: Calculates precision, recall, and f1-score for each species class.
*   `df_class_report.to_csv(...)`: Exports the resulting metrics to Drive for inclusion in the final report.

#### 5. Expected Output
*   A **Confusion Matrix** grid where the diagonal represents correct predictions. Numbers outside the diagonal will reveal taxonomic confusion (e.g., model confusing *E. coli* with *Salmonella* because both are closely related Gram-negative enterobacteria with highly similar GC profiles).
*   A **Classification Report** detailing which species were identified with the highest reliability and which ones were the hardest to predict at short lengths.

#### 6. Biological Significance
This evaluation lets us see if closely related bacteria are harder to differentiate at short sequence lengths. For instance, *Escherichia coli* and *Salmonella enterica* are both members of the Enterobacteriaceae family, sharing high genetic similarity. At only 100bp, their base compositions are often indistinguishable, which should show up as a high confusion rate between them on the matrix!
